```python
mybio = bd.Database('biosphere3')
eidb = bd.Database('ecoinvent-391-cutoff')

print(type(mybio))
print(len(mybio))

print(type(eidb))
print(len(eidb))
```

```python
elecwind =[act for act in eidb if 'electricity production' in act['name'] and 'wind, 1-3MW turbine, offshore' in act['name'] and 'BE' in act['location']][0]
elecwind.as_dict()
```

```python
for x in elecwind.exchanges():
    print(x)
```

```python
[m for m in bd.methods if 'EF v3.1' in str(m) and 'no LT' not in str(m) and 'EN1' not in str(m)]
```

```python
climate_change = [m for m in bd.methods if 'EF v3.1' in str(m) and 'no LT' not in str(m) and 'EN1' not in str(m) and 'climate change' in str(m) and 'biogenic' not in str(m) and 'fossil' not in str(m) and 'land use' not in str(m)][0]
```

```python
windlca = bc.LCA({elecwind:1}, climate_change)
windlca.lci()
windlca.lcia()
windlca.score
```

```python
print("The Greenhouse Gas emissions for 1 kWh electricity produced by a wind turbine in BE is:", windlca.score*1000, " gCO2eq/kWh")
```
